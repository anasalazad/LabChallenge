```python
import pandas as pd
import time # to add time wrapper for latency call
```

```python
df = pd.read_csv("dataset/student-mat.csv")
df = df.head(20) # keep it small for the lab
```

## Testing small dataset

```python
# Using ollama (local)
def get_gender_count_ollama(gender):
    start = time.time()
    values = list(gender)
    response = ollama.chat(
        model='llama3.2:1b',
        messages=[
            {'role': 'user', 'content': f"Here is the 'sex' column from a dataset: {values}\n\nHow many Male and Female are in this dataset?"}
        ]
    )
    end = time.time()

    latency = end - start
    print(f"Latency: {latency:.2f}s")

    return response['message']['content'].strip()
```

```python
gender_count_ollama = get_gender_count_ollama(df['sex'])
```

```python
print(gender_count_ollama)
```

```python
# Google (cloud)
def get_gender_count_gemini(gender):
    start = time.time()
    values = list(gender)
    response = client.models.generate_content(
        model="gemini-3.5-flash",
        contents=f"Here is the 'sex' column from a dataset: {values}\n\nHow many Male and Female are in this dataset?"
    )
    end = time.time()
    latency = end - start
    print(f"Latency: {latency:.2f}s")

    return response.text.strip()
```

```python
gender_count_gemini = get_gender_count_gemini(df['sex'])
```

```python
print(gender_count_gemini)
```

## Testing model Reasoning

```python
# Ollama (Reasoning)
def get_apples_reason_ollama():
    start = time.time()
    response = ollama.chat(
        model='llama3.2:1b',
        messages=[
            {'role': 'user', 'content': "Think step by step, then give your final answer."
            "\n\nQuestion: A store has 3 boxes of 12 apples each. 5 apples are rotten. "
            "How many apples remain?"}
        ]
    )
    end = time.time()
    latency = end - start
    print(f"Latency: {latency:.2f}s")

    return response['message']['content'].strip()
```

```python
apples_ollama = get_apples_reason_ollama()
```

```python
print(apples_ollama)
```

```python
# Ollama (Non-Reasoning)
def get_apples_answer_ollama():
    start = time.time()
    response = ollama.chat(
        model='llama3.2:1b',
        messages=[
            {'role': 'user', 'content': "Question: A store has 3 boxes of 12 apples each. 5 apples are rotten. "
            "How many apples remain?"}
        ]
    )
    end = time.time()
    latency = end - start
    print(f"Latency: {latency:.2f}s")

    return response['message']['content'].strip()
```

```python
apples_answer_ollama = get_apples_answer_ollama()
```

```python
print(apples_answer_ollama)
```

## Using GradioUI for interactive chat

```python
import gradio as gr
```

```python
def chat_with_ollama(message, history):
    response = ollama.chat(
        model='llama3.2:1b',
        messages=[
            {'role': 'user', 'content': message}
        ]
    )
    return response['message']['content']
```

```python
demo = gr.ChatInterface(
    fn=chat_with_ollama,
    title="Week 1: Local LLM Chat (Ollama)",
    description="Chatting with llama3.2:1b running via Ollama"
)

demo.launch()
```

# Checkpoint Questions
## What's the practical difference in setup complexity between local and cloud?
The difference is that cloud would require you to go to the official website and download it. Then it is ready to use.

For local, you would need to install it in a IDE using the command prompt. Then you would need a few lines of code to test if it works.

|Local|Cloud|
|-----|-----|
|Hosted on your local computer|Hosted on a company super computers|
|Uses personal computer processor and memory|Accessible via the internet|
|Has smaller parameters to fit onto the computer chip|Has many parameters, giving it maximum knowledge|
|Provides data privacy and no subscription cost|Higher speed, accuracy and solving most complex problems|
|Output quality is poor and slower as it is using your CPU|Data must be sent over the internet, raising privacy and security concern|
|Handle highly sensitive tasks like drafting confidential legal documents or private note taking|Handle deep market research, advanced coding or generated content|

## Where did the two models disagree, and why might that happen?
During the counting task for the different gender, the Ollama model had the total count of 17 and the Gemini model had the total count of 20. The Gemini model had the correct answer. This could be concluded that the Ollama model (local) has lower parameters to handle math problems.

For the reasoning task, both models explained their reasoning process, but the Gemini model also guides the user how to solve the problem manually.
