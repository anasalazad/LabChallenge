## 3. Type conversion
### Implicit – Python promotes `int` → `float` automatically so no data is lost

```python
my_int = 5
my_float = 2.15
add_int_float = my_int + my_float

print("Datatype of my_int is: ", type(my_int))
print("Datatype of my_float is: ", type(my_float))

print("Value of add_int_float is: ", add_int_float)
print("Datatype of add_int_float:", type(add_int_float))
```

### Explicit (type casting) – we convert ourselves with `int()`, `float()`, `str()`

```python
num_int = 123
num_str = "456"

print("Data type of num_int: ", type(num_int))
print("Data type of num_str before Type Casting: ", type(num_str))

num_str = int(num_str)
print("Data type of num_str after Type Casting: ", type(num_str))

num_sum = num_int + num_str

print("Sum of num_int and num_str: ", num_sum)
print("Data type of the sum: ", type(num_sum))
```

**My try:** what happens if I *don't* cast? Python refuses to add an `int` and a `str` (no implicit conversion there), so I caught the error to see the message.

```python
try:
    print(123 + "456")
except TypeError as e:
    print("TypeError:", e)

print(str(123) + "456")    # string concatenation instead -> '123456'
print(float("3.5") * 2)    # string -> float works too
print(int(7.9))            # int() chops the decimals off (doesn't round!)
```

## 6. While loop
The condition is checked **before** each iteration; loop stops when it becomes `False`.

```python
# Program to add natural numbers up to sum = 1+2+3+...+n

n = 10

# initialize sum and counter
sum = 0
i = 1

while i <= n:
    sum = sum + i
    i = i+1    # update counter

# print the sum
print("The sum is", sum)
```

## 8. The `range()` function
`range(start, stop, step)` – `stop` is **exclusive**, `step` defaults to 1.

```python
# Program to iterate through a list of integers

n = 10

for i in range(n):
    print(i, end=" ")   # To print without starting a new line
```

```python
# Program to iterate through a list using indexing

genre = ['pop', 'rock', 'jazz']

# iterate over the list using index
for i in range(len(genre)):
    print("I like", genre[i])
```

**My try:** `start`/`step` arguments, counting backwards, and `enumerate()` which gives the index *and* the item (cleaner than `range(len(...))`).

```python
print(list(range(2, 20, 3)))
print(list(range(10, 0, -2)))
for i, g in enumerate(genre):
    print(i, g)
```
