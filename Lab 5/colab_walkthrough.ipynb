TPU check from the lab sheet (switch the runtime to TPU first if you want to see *Connected to TPU*):

```python
import os

if 'COLAB_TPU_ADDR' not in os.environ:
    print('Not connected to TPU')
else:
    print("Connected to TPU")
```

*Note to self:* the newer Colab "TPU v2-8" runtimes don't always set `COLAB_TPU_ADDR` any more, so the cell above can say *Not connected* even on a TPU runtime. The cell below is a more reliable check.

```python
import tensorflow as tf
try:
    tpus = tf.config.list_logical_devices('TPU')
    print("TPU devices:", tpus if tpus else "none")
except Exception as e:
    print("No TPU:", e)
```

### Uploading a file to session storage
Either the *Upload to session storage* button in the Files pane, or from code:

```python
from google.colab import files
uploaded = files.upload()          # opens a file picker - pick any small file
for name, data in uploaded.items():
    print(f"uploaded {name} ({len(data)} bytes)")
```

### Mounting Google Drive
