```python
from smolagents import InferenceClientModel, LogLevel
```

```python
# Setup cloud vs local agents
# Use verbosity_level=LogLevel.ERROR to suppress log

cloud_model = InferenceClientModel(
    model_id="meta-llama/Llama-3.3-70B-Instruct",
    # token comes from the HF_TOKEN environment variable - not stored in the notebook
)

cloud_agent = CodeAgent(
    tools=[DuckDuckGoSearchTool()],
    model=cloud_model,
    verbosity_level=LogLevel.ERROR
)

local_model = TransformersModel(model_id="Qwen/Qwen2.5-1.5B-Instruct")
local_agent = CodeAgent(
    tools=[DuckDuckGoSearchTool()],
    model=local_model,
    verbosity_level=LogLevel.ERROR
)
```

```python
import time
```

```python
def run_and_time(agent, task):
    start = time.time()
    result = agent.run(task)
    elapsed = time.time() - start
    return result, elapsed
```

```python
task = "What is the current weather in Melbourne?"
```

```python
cloud_result, cloud_time = run_and_time(cloud_agent, task)
local_result, local_time = run_and_time(local_agent, task)

print(f"Cloud: {cloud_time:.2f}s\n{cloud_result}\n")
print(f"Local: {local_time:.2f}s\n{local_result}")
```

## Class notes

agent architecture & design patterns

1. what makes an AI systems "Agentic"?

a standard llm mainly responses prompts, wheareas the agent work toward a goal using 5 steps
reasoning, planning, memory handling, tools using and repeated actions

core loop: observe -> think -> plan -> act -> repeat until the goal is met

2. core agent architecture:
llm: reasoning engine (like a brain)
memory: database store context, history conversations/outputs, and the knowledge
planner: breaks a goal into smaller tasks
tool executor: calls apis, databases, code, web, email,...
environment the world the agent observes and acts on
refection: checks results and improves or replans

3. Memory:
context window != memory: the context window is temporary and limited, memory can persist, be searched, updated and reused across different sessions

4 memory types:
+ working(now)
+ short-term(current conversation)
+ long-term (across different sessions)
+ procedural (how to perform tasks)
memory lifecycle: observe -> store -> index -> retrieve -> use -> update -> forget
important principle: **retrieval quality** is more important than simply storing more data

4. 3 core design patterns
tools used: route subtasks to external tools such as web search, scraping, Python executor, databases, APIs or email
reflection: generate -> Critique -> Refine. Prefer tests, compilers or other external feedback when possible
Planning: decompose the complex goal into smaller tasks, execute them, observe the results and replan if needed

5. autonomy levels of agents
- L0: prompt-based
- L1: assistive
- L2: supervised task-based
- L3: conditional operator
- L4: high autonomy
- L5 :full autonomy
